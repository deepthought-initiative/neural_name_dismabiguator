```python
author_names = author_references[['author', 'author2']]
affiliations = author_references[['aff', 'aff2']]
```

```python
c2v_model = chars2vec.load_model('eng_50')

author1 = author_names.author.to_list()
author2 = author_names.author2.to_list()
aff1 = affiliations.aff.to_list()
aff2 = affiliations.aff2.to_list()
```

```python
author1_embeddings = c2v_model.vectorize_words(author1)
```

```python
author2_embeddings = c2v_model.vectorize_words(author2)
```

```python
aff1_embeddings = c2v_model.vectorize_words(aff1)
```

```python
aff2_embeddings = c2v_model.vectorize_words(aff2)
```

```python
np.save("author1_embeddings",author1_embeddings)
np.save("author2_embeddings",author2_embeddings)
np.save("aff1_embeddings",aff1_embeddings)
np.save("aff2_embeddings",aff2_embeddings)
```

```python
def normalize_vector(vector):
    # Calculate the Euclidean norm (magnitude) of the vector
    magnitude = np.sqrt(sum(x ** 2 for x in vector))

    # Normalize the vector by dividing each element by its magnitude
    normalized_vector = [x / magnitude for x in vector]

    return normalized_vector

# Example usage:
vector = [3, 4, 5]
normalized_vector = normalize_vector(vector)
print(normalized_vector)
```

```python
norm_author1 = np.array(normalize_vector(author1_embeddings))
norm_author2 = np.array(normalize_vector(author2_embeddings))
```

```python
norm_author1.tofile("authors")
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def calculate_cosine_similarity(array1, array2):
    """
    Calculate cosine similarity between inner arrays of two 2D arrays.

    Parameters:
    - array1: First 2D array
    - array2: Second 2D array (should have the same shape as array1)

    Returns:
    - cosine_similarities: 1D array containing the cosine similarities
      between corresponding inner arrays of array1 and array2.
    """

    if array1.shape != array2.shape:
        raise ValueError("Input arrays must have the same shape")

    num_rows = array1.shape[0]
    cosine_similarities = np.zeros(num_rows)

    for i in range(num_rows):
        cosine_similarities[i] = cosine_similarity([array1[i]], [array2[i]])[0, 0]

    return cosine_similarities

# Example usage:
similarities = calculate_cosine_similarity(norm_author1, norm_author2)
print(similarities)
```
